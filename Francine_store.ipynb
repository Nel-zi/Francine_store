### Extraction Layer

```python
# Configure Selenium chrome Driver options
options = Options()
options.headless = True
service = Service(executable_path = r'chromedriver-win64\chromedriver.exe')

# Initialize the webdriver
driver = webdriver.Chrome(service = service, options = options)

# Lists to store extracted data
product_names = []
prices = []
store_names = []
store_links = [] 
shipping_prices = []
extra_discounts = []
items_sold = []
original_prices = []
shipping_free_statuses = []


# Scrapping the pages
total_pages = range(1, 2)

for page in total_pages:
    url = f'https://www.aliexpress.com/w/wholesale-laptop.html?page={page}&g=y&SearchText=laptop'
    driver.get(url)
    time.sleep(10)

    # Now we use Beautifulsoup to parse the loaded page_source
    soup = BeautifulSoup(driver.page_source, 'html.parser')

    # Adjust the class selectors based on the current website structure
    laptops = soup.find_all('div', class_ = 'list--gallery--C2f2tvm search-item-card-wrapper-gallery')

    for laptop in laptops:
        # Product_name
        product_name = laptop.find('h3', class_ ='multi--titleText--nXeOvyr').text
        

        # Price
        price = laptop.find('div', class_ ='multi--price-sale--U-S0jtj').text.replace('NGN', '').replace(',', '')

       
        # Store name
        store_n = laptop.find('a', class_ ='cards--storeLink--XkKUQFS')
        store_name = store_n.text if store_n else ''

        
        #Store link
        try:
            store_link = laptop.find('a', class_ ='cards--storeLink--XkKUQFS')['href']

        except KeyError as e:
            store_link = ''
            print(f"KeyError occurred in store_link: {e}")

        except Exception as e:
            store_link = ''
            print(f"Error Type in store_link: {type(e)}")        # Error class
            print(f"Error Arguments in store_link: {e.args}")    # Error message and details

        
        # Shipping price
        try:
            shipping_p = laptop.find('span', class_ = 'tag--text--1BSEXVh tag--textStyle--3dc7wLU multi--serviceStyle--1Z6RxQ4')
            shipping_price = shipping_p.text if shipping_p else ''
        except AttributeError as e:
            shipping_price = ''
            print(f"KeyError occurred in shipping_price: {e}")

        except Exception as e:
            shipping_price = ''
            print(f"Error Type in shipping_price: {type(e)}")        # Error class
            print(f"Error Arguments in shipping_price: {e.args}")    # Error message and details

        
        # Extra discount
        try:
            extra_dis = laptop.find('span', class_ = 'tag--text--1BSEXVh tag--textStyle--3dc7wLU multi--superStyle--1jUmObG')
            extra_discount = extra_dis.text if extra_dis else ''

        except AttributeError as e:
            extra_discount = ''
            print(f"KeyError occurred in extra_discount: {e}")
            
        except Exception as e:
            extra_discount = ''
            print(f"Error Type in extra_discount: {type(e)}")        # Error class
            print(f"Error Arguments in extra_discount: {e.args}")    # Error message and details

        
        # Item sold
        try:
            item_sold = laptop.find('span', class_ = 'multi--trade--Ktbl2jB')
            item_sold_split = item_sold.text.split()[0] if item_sold else ''
        
        except AttributeError as e:
            item_sold = ''
            print(f"KeyError occurred in item_sold: {e}")

        except Exception as e:
            item_sold = ''
            print(f"Error Type in item_sold: {type(e)}")        # Error class
            print(f"Error Arguments in item_sold: {e.args}")    # Error message and details

        
        # Original_prices
        try:
            original_price = laptop.find('div', class_ = 'multi--price-original--1zEQqOK')
            old_price = original_price.text.strip() if original_price else ''

        except Exception as e:
            original_price = ''
            print(f"Error Type in original_price: {type(e)}")        # Error class
            print(f"Error Arguments in original_price: {e.args}")    # Error message and details

        
        # Shipping free statuses
        try:
            shipping_statuses = laptop.find('div', class_ = 'multi--serviceContainer--3vRdzWN')
            ship_status = shipping_statuses.text.strip() if shipping_statuses else ''

        except Exception as e:
            shipping_statuses = ''
            print(f"Error Type in shipping_statuses: {type(e)}")        # Error class
            print(f"Error Arguments in shipping_statuses: {e.args}")    # Error message and details


        # Append data to list
        product_names.append(product_name)
        prices.append(price)
        store_names.append(store_name)
        store_links.append(store_link)
        shipping_prices.append(shipping_price)
        extra_discounts.append(extra_discount)
        items_sold.append(item_sold_split)
        original_prices.append(old_price)
        shipping_free_statuses.append(ship_status)


# Once all the pages information are extracted, quit the driver
driver.quit()   
        
```

#### Create a dataframe

```python
data = {
    'Product_name' : product_names,
    'Price' : prices,
    'Store_name' : store_names,
    'Store_link' : store_links,
    'Shipping_price' : shipping_prices,
    'Extra_discount' : extra_discounts,
    'Items_sold' : items_sold,
    'Original_price' : original_prices,
    'Shipping_free_statuse' : shipping_free_statuses
}

francine_df = pd.DataFrame(data)
```

### Data Transformation

###### Creating tables

```python
# Product table
product_columns = ['Product_name', 'Price', 'Store_name', 'Store_link']
products_table = francine_df[product_columns].copy().drop_duplicates().reset_index(drop=True)

# Creating a product id using a dedicated numbering system
products_table['product_id'] = range(7, len(products_table) + 7) # Sequential numbering starting at 7
products_table = products_table[['product_id', 'Product_name', 'Price', 'Store_name', 'Store_link']]

products_table.head()
```

```python
# Discount table
discount_column = ['Original_price', 'Extra_discount']
discounts_table = francine_df[discount_column].copy().drop_duplicates().reset_index(drop=True)

# Adding the product_id from the products_table
to_join_product_id = products_table['product_id'].copy().reset_index(drop=True)
discounts_table['product_id'] = to_join_product_id

# Creating a discounts id using a dedicated numbering system
discounts_table['discount_id'] = range(44, len(discounts_table) + 44) # Sequential numbering starting at 44
discounts_table = discounts_table[['discount_id', 'product_id', 'Original_price', 'Extra_discount']]

# Removing currency symbols and commas then we convert to float
discounts_table['Original_price'] = discounts_table['Original_price'].str.replace('NGN', '').str.replace(',', '').str.extract(r'(\d+)').fillna(0).astype(float)
                                                                                               
# extract discount percent value and coin percent value as integer
discounts_table['Extra_discount'] = discounts_table['Extra_discount'].str.extract(r'(\d+)%').fillna(0).astype(float)



discounts_table
```

```python
# Sales table
sales_column = ['Items_sold']
sales_table = francine_df[sales_column].copy().drop_duplicates().reset_index(drop=True)

# Adding the product_id from the products_table
columns_to_join = products_table['product_id'].copy().reset_index(drop=True)
sales_table['product_id'] = columns_to_join


# Creating a Sales id using a dedicated numbering system
sales_table['sales_id'] = range(131, len(sales_table) + 131) # Sequential numbering starting at 131
sales_table = sales_table[['sales_id', 'product_id', 'Items_sold']]

sales_table
```

```python
# Shipping table
shipping_column = ['Shipping_price']
shipping_table = francine_df[shipping_column].copy().drop_duplicates().reset_index(drop=True)

# Adding the product_id from the products_table
joining_product_id = products_table['product_id'].copy().reset_index(drop=True)
shipping_table['product_id'] = joining_product_id

# Creating a Sales id using a dedicated numbering system
shipping_table['shipping_id'] = range(250, len(shipping_table) + 250) # Sequential numbering starting at 250
shipping_table = shipping_table[['shipping_id', 'product_id', 'Shipping_price']]

shipping_table.fillna({'Shipping_price' : 0.0}, inplace=True)

shipping_table
```

### Saving the tables to CSV

```python
products_table.to_csv('products_table.csv', index=False)
sales_table.to_csv('sales_table.csv', index=False)
discounts_table.to_csv('discounts_table.csv', index=False)
shipping_table.to_csv('shipping_table.csv', index=False)
```

### Loading Layer

```python
# Load environment variables from the .env file
load_dotenv()

db_params = {
    'username' : os.getenv('DB_USER'),
    'password' : os.getenv('PASSWORD'),
    'host' : os.getenv('HOST'),
    'port': int(os.getenv('PORT')),  # Ensure PORT is an integer
    'database' : os.getenv('DATABASE')
}

# define the database connection url
db_url = f"postgresql://{db_params['username']}:{db_params['password']}@{db_params['host']}:{db_params['port']}/{db_params['database']}"


# create the database engine with the db url specified
engine = create_engine(db_url)

# connect to postgreSQL server
with engine.connect() as connection:
    # Create table and load the data
    products_table.to_sql('products_table', connection, index = False, if_exists = 'replace')
    discounts_table.to_sql('discounts_table', connection, index = False, if_exists = 'replace')
    sales_table.to_sql('sales_table', connection, index = False, if_exists = 'replace')
    shipping_table.to_sql('shipping_table', connection, index = False, if_exists = 'replace')

print('Database, table and data loaded successfully')


```
